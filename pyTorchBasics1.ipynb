#3. Reduction Operation

```python
e = torch.randint(size=(2,3), low=0, high=10, dtype = torch.float32)
e
```

```python
# sum
total_sum = torch.sum(e)

# sum along columns
total_column_sum = torch.sum(e, dim=0)

# sum along rows
total_row_sum = torch.sum(e, dim=1)

print(total_sum)
print(total_column_sum)
print(total_row_sum)
```

```python
# mean
total_mean = torch.mean(e)

# mean along columns
total_col_mean = torch.mean(e, dim=0)

# mean along rows
total_row_mean = torch.mean(e, dim=1)

print(total_mean)
print(total_col_mean)
print(total_row_mean)
```

```python
# meadian
total_median = torch.median(e)

# median along col
total_col_median = torch.median(e, dim=0)

# median along row
total_row_median = torch.median(e, dim=1)

print("Total median -> ",total_median)
print("Total column median -> ",total_col_median)
print("Total row median -> ",total_row_median)
```

```python
# max and min
total_max = torch.max(e)
print(total_max)

# max along col
total_col_max = torch.max(e,dim=0)
print(total_col_max)

# max along row
total_row_max = torch.max(e,dim=1)
print(total_row_max)

# min
total_min = torch.min(e)
print("Total min ->", total_min)
```

```python
# product
total_prod = torch.prod(e)
print(total_prod)
```

```python
# standard deviation
total_std = torch.std(e)
print(total_std)
```

```python
# variance
total_var = torch.var(e)
print(total_var)
```

```python
# argmax
torch.argmax(e)
```

```python
# argmin
torch.argmin(e)
```

#4. Matrix Operations

```python
f = torch.randint(size=(2,3), low=0, high=10, dtype=torch.float32)
g = torch.randint(size=(3,2), low = 0, high = 10, dtype=torch.float32)

print(f)
print(g)
```

```python
# matrix multiplication
torch.matmul(f,g)
```

```python
vector1 = torch.tensor([1,2])
vector2 = torch.tensor([3,4])

# dot product
torch.dot(vector1, vector2)
```

```python
# transpose
torch.transpose(f,0,1)
```
